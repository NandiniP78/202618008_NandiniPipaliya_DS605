##### Task 2 - Vectorized Arithmetic and Linear Algebra

```python
# 1. Matrix Operations: Addition, Element-wise, and Matrix Product
# Creating two 2x2 matrices
A = np.array([
    [4.0, 7.0],
    [2.0, 6.0]
])

B = np.array([
    [1.0, 3.0],
    [5.0, 2.0]
])

print("=== Matrix A ===")
print(A)
print("\n=== Matrix B ===")
print(B)

# (a) Addition
add_result = A + B
print("\n1. Matrix Addition (A + B):")
print(add_result)

# (b) Element-wise Multiplication
elem_mult = A * B
print("\n2. Element-wise Multiplication (A * B):")
print(elem_mult)

# (c) Matrix Multiplication
matmul_result = A @ B  # Identical to np.matmul(A, B)
print("\n3. Matrix Multiplication (A @ B):")
print(matmul_result)
```

```python
# 2. Transpose and Determinant of a Square Matrix
print("=== Square Matrix A: Transpose & Determinant ===")

# Transpose
A_transpose = A.T
print("\nTranspose (A.T):")
print(A_transpose)

# Determinant
det_A = np.linalg.det(A)
print(f"\nDeterminant det(A): {det_A:.4f}")
```

```python
# 3. Inverse and Verification
print("=== Invertibility & Identity Verification ===")

# Check if invertible (determinant is non-zero)
if not np.isclose(det_A, 0.0):
    # Compute inverse
    A_inv = np.linalg.inv(A)
    print("\nInverse Matrix (A^-1):")
    print(A_inv)

    # Compute A @ A^-1
    product = A @ A_inv
    print("\nResult of A @ A^-1:")
    print(product)

    # Create identity matrix of matching shape
    identity = np.eye(A.shape[0])

    # Verify equivalence using np.allclose
    is_identity = np.allclose(product, identity)
    print(f"\nDoes A @ A^-1 equal the Identity Matrix (I)? -> {is_identity}")
else:
    print("\nMatrix A is singular (determinant is 0) and cannot be inverted.")
```

##### Task 3 - Normal Distribution and Histogram 

```python
# 1. Generate Random Normal Samples
# Set seed for reproducibility
rng = np.random.default_rng(seed=42)

# Population (theoretical) parameters
mu_chosen = 50.0
sigma_chosen = 10.0
sample_size = 1000

# Generate 1,000 values from N(mu=50, sigma^2=100)
samples = rng.normal(loc=mu_chosen, scale=sigma_chosen, size=sample_size)
```

```python
# 2. Compute Sample Statistics & Comparison
sample_mean = np.mean(samples)
# ddof=1 computes the unbiased sample standard deviation (divides by N-1)
sample_std = np.std(samples, ddof=1)

print("=== Parameter vs. Sample Comparison ===")
print(f"{'Metric':<25} | {'Theoretical':<12} | {'Sample Estimate':<15} | {'Difference':<10}")
print("-" * 70)
print(f"{'Mean (μ)':<25} | {mu_chosen:<12.4f} | {sample_mean:<15.4f} | {abs(sample_mean - mu_chosen):<10.4f}")
print(f"{'Standard Deviation (σ)':<25} | {sigma_chosen:<12.4f} | {sample_std:<15.4f} | {abs(sample_std - sigma_chosen):<10.4f}")
```

```python
# 3. Plot Histogram with Theoretical Probability Density Function (PDF)
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5.5))

# Plot normalized histogram (density=True normalizes bin areas to sum to 1)
count, bins, ignored = plt.hist(
    samples, 
    bins=30, 
    density=True, 
    alpha=0.6, 
    color='royalblue', 
    edgecolor='black', 
    label='Sample Histogram'
)

# Overlay theoretical normal distribution PDF curve
pdf = (1 / (sigma_chosen * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((bins - mu_chosen) / sigma_chosen) ** 2)
plt.plot(bins, pdf, color='crimson', linewidth=2.5, label=f'Theoretical PDF ($\mu={mu_chosen}, \sigma={sigma_chosen}$)')

# Reference vertical lines
plt.axvline(sample_mean, color='navy', linestyle='--', linewidth=1.8, label=f'Sample Mean ({sample_mean:.2f})')
plt.axvline(mu_chosen, color='darkred', linestyle=':', linewidth=1.8, label=f'True Mean ({mu_chosen:.2f})')

# Labels and Styling
plt.title(f'Normal Distribution ($N={sample_size}$)', fontsize=14, pad=12)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.legend(loc='upper right', frameon=True)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

##### Task 7 - Missing Values and Fare Outliers 


```python
print("1. Missing Value Analysis")
missing_count = df.isnull().sum()
missing_pct = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing (%)': missing_pct.round(2)
})
print(missing_summary)

# Plotting Missing Values Bar Chart
plt.figure(figsize=(8, 4.5))
missing_count.plot(kind='bar', color='steelblue', edgecolor='black')
plt.title('Missing Value Count per Column', fontsize=13, pad=10)
plt.xlabel('Columns', fontsize=11)
plt.ylabel('Number of Missing Values', fontsize=11)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
print("2. Mean Imputation on Age")
df_mean_demo = df.copy()
print(f"Missing Age count BEFORE: {df_mean_demo['Age'].isnull().sum()}")

# Fill with mean
mean_age = df_mean_demo['Age'].mean()
df_mean_demo['Age'] = df_mean_demo['Age'].fillna(mean_age)

print(f"Calculated Mean Age:     {mean_age:.2f}")
print(f"Missing Age count AFTER:  {df_mean_demo['Age'].isnull().sum()}")
```

```python
print("3. Comparison of 4 Imputation Techniques")
# 1. Mean
mean_val = df['Age'].mean()
age_mean = df['Age'].fillna(mean_val)

# 2. Median
median_val = df['Age'].median()
age_median = df['Age'].fillna(median_val)

# 3. Mode
mode_val = df['Age'].mode()[0]
age_mode = df['Age'].fillna(mode_val)

# 4. Random Sampling from non-null distribution
rng = np.random.default_rng(seed=42)
non_null_ages = df['Age'].dropna().values
age_random = df['Age'].apply(
    lambda x: rng.choice(non_null_ages) if pd.isna(x) else x
)

# Demonstration on rows with missing values
missing_indices = df[df['Age'].isna()].index
imputation_comparison = pd.DataFrame({
    'Row Index': missing_indices,
    'Mean Fill': [mean_val] * len(missing_indices),
    'Median Fill': [median_val] * len(missing_indices),
    'Mode Fill': [mode_val] * len(missing_indices),
    'Random Fill': age_random.loc[missing_indices].values
})
print(imputation_comparison.round(2))
```

```python
print("4. Fare Outlier Detection (IQR Method)")
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]

print(f"Q1 (25th percentile) : {Q1:.4f}")
print(f"Q3 (75th percentile) : {Q3:.4f}")
print(f"IQR (Q3 - Q1)        : {IQR:.4f}")
print(f"Lower Bound (Q1-1.5x): {lower_bound:.4f}")
print(f"Upper Bound (Q3+1.5x): {upper_bound:.4f}")
print(f"Total Outliers Found : {len(outliers)}")

if len(outliers) > 0:
    print("\nOutlier Rows:")
    print(outliers[['PassengerId', 'Pclass', 'Fare']])
```

##### Task 8 - Features and Pivot Table 


```python
print("1. Feature Engineering (FamilySize & IsAlone)")
# Calculate FamilySize
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# Create binary IsAlone flag (1 if alone, 0 otherwise)
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

# Display sample output
print(df[['PassengerId', 'Sex', 'Pclass', 'SibSp', 'Parch', 'FamilySize', 'IsAlone', 'Survived']].head(8))
```

```python
print("2. Pivot Table & Identifying Extremes")
# Create pivot table
pivot_survival = df.pivot_table(
    index='Sex', 
    columns='Pclass', 
    values='Survived', 
    aggfunc='mean'
)

print("--- Survival Rate Pivot Table ---")
print(pivot_survival.map(lambda x: f"{x:.2%}" if pd.notnull(x) else "N/A"))

# Find Highest and Lowest Survival Groups
# Unstack converts the 2D pivot table to a 1D Series indexed by (Pclass, Sex)
flat_rates = pivot_survival.unstack().dropna()

max_rate = flat_rates.max()
min_rate = flat_rates.min()

highest_groups = flat_rates[flat_rates == max_rate].index.tolist()
lowest_groups = flat_rates[flat_rates == min_rate].index.tolist()

print("\n--- Summary Findings ---")
print(f"Highest Survival Rate ({max_rate:.2%}):")
for pclass, sex in highest_groups:
    print(f"  • Class {pclass} {sex.capitalize()}")

print(f"\nLowest Survival Rate ({min_rate:.2%}):")
for pclass, sex in lowest_groups:
    print(f"  • Class {pclass} {sex.capitalize()}")
```
